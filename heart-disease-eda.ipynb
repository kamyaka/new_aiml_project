```python
# Import the libraries needed for data handling (pandas) and visualization (seaborn, matplotlib)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Q1: Load the heart disease dataset from a CSV file and display the first 5 rows
df = pd.read_csv('heart.csv.xls')
df.head(5)
```

```python
# Q2: Display the shape of the dataset (number of rows and columns)
df.shape
```

```python
# Q3: Display the data type of each column
df.dtypes
```

```python
# Q4: Check the number of missing (NaN) values in each column
df.isna().sum()
```

```python
# Q5: Check for duplicate rows, then remove them and confirm none remain
df.duplicated().sum()
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
# Q6: Display summary statistics (count, mean, std, min, max, quartiles) for numeric columns
df.describe()
```

```python
# Q7: Display the number of records for each class in the 'target' column
df['target'].value_counts()
```

```python
# Q8: Plot a histogram to visualize the distribution of the 'target' column
sns.histplot(data = df, x = 'target')
plt.show()
```

```python
# Q9: Plot a count plot of 'target' grouped by 'sex' to compare classes across genders
sns.countplot(df,x = 'target', hue= 'sex')
plt.show()
```

```python
# Q10: Plot a histogram to visualize the distribution of patients' ages
sns.histplot(df, x = 'age')
plt.show()
```

```python
# Q11: Plot a boxplot of 'chol' (cholesterol) to check its distribution and outliers
sns.boxplot(df, x = 'chol')
plt.show()
```

```python
# Q12: Compute and plot the average cholesterol level for each target class
avgChol = df.groupby('target')['chol'].mean()
sns.barplot(x = avgChol.index, y = avgChol.values)
plt.xlabel("Target")
plt.ylabel("Avg Chol Level")
plt.show()
```

```python
# Q13: Plot a scatter plot of 'thalach' (max heart rate) against 'target'
sns.scatterplot(data=df, x = df['thalach'], y = df['target'])
plt.show()
```

```python
# Q14: Plot a heatmap of the correlation matrix between all numeric features
corrMatrix = df.select_dtypes('number').corr()
sns.heatmap(corrMatrix,linewidths=0.7,cmap='Reds')
plt.show()
```
